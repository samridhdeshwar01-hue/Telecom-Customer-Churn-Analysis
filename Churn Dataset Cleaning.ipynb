**Replacing inconsistent missing values with a standard NaN (This helps standardize missing values in a DataFrame because pandas recognizes NaN as a missing value.)**

```python
df.replace(['N/A', 'NULL', ''], np.nan, inplace = True)
```

**Stripping extra whitespace from text (string) columns**

**We will remove unwanted leading and trailing spaces from text values so that data is clean and consistent.**

```python
df = df.apply(lambda x:x.str.strip() if x.dtype == "str" else x)
```

**Standardizing values to title case**

```python
df["Customer_Name"] = df["Customer_Name"].str.title()
df["State"] = df["State"].str.title()
df["City"] = df["City"].str.title()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Churn"] = df["Churn"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Gender"] = df["Gender"].str.title()
df["Payment_Method"] = df["Payment_Method"].str.title()
df["Internet_Service"] = df["Internet_Service"].str.title()
df["Tech_Support"] = df["Tech_Support"].str.title()
df["Senior_Citizen"] = df["Senior_Citizen"].str.title()
df["Dependents"] = df["Dependents"].str.title()
```

**Checking if any lower case value remains (any 2 random columns)**

```python
df["Subscription_Type"].value_counts()
```

```python
df["Tech_Support"].value_counts()
```

**Removing invalid ages above 100 years**

```python
df = df[(df['Age'] >=18) & (df['Age'] <=100)]
```

**Removing Negative values in charges column**

```python
df = df[df["Monthly_Charges"]>= 0]
df = df[df["Total_Charges"]>= 0]
```

```python
df.shape
```

**Converting 'Last_Interaction_Date' to a proper datetime format and convert invalid values to NaT (missing date/time)**

**We are doing this because dates stored as plain text cannot be reliably sorted, filtered, compared, or used for date calculations; converting them to datetime lets pandas perform these operations correctly**

**errors="coerce" prevents invalid dates from causing errors by converting them to NaT.**

```python
df["Last_Interaction_Date"] = pd.to_datetime(df["Last_Interaction_Date"], errors = 'coerce')
```

**Filling missing values**

```python
df["Tech_Support"]=df["Tech_Support"].fillna("No")

df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")

df["Monthly_Charges"]=df["Monthly_Charges"].fillna(df['Monthly_Charges'].mean())

df["Tenure_Months"]=df["Tenure_Months"].fillna(df['Tenure_Months'].mean())

df['Age'] = df['Age'].fillna(df['Age'].mean())
```

**Starting Feature Engineering here since the data is now clean, and we need to derive business-relevant metrics that the raw columns don't directly provide. We will be answering our questions here:**

First we wil find out Customer Lifetime Value (how much a customer pays 

```python
df['Customer_Value'] = df["Monthly_Charges"] *df["Tenure_Months"]
```

**Let's make a column for monthly revenue which will be same as Monthly_Charges.**

```python
df["Monthly_Revenue"] = df["Monthly_Charges"]
```

**We wil now group the custromers based on their Tenure (0-12, 13-24, 25-48, 49-72)**

```python
bins=[0,12,24,48,72]
label = labels = ["0-12 Months","13-24 Months","25-48 Months","49-72 Months"]
df["Tenure_Group"] = pd.cut(df["Tenure_Months"], bins = bins, labels = label)
```

**Adding a senior flag column using numpy where functions.**

```python
df["Senior_Flag"]=np.where(df["Age"]>=60,"Senior","Adult")
```

**Adding a churn column now.**

```python
df["Churn_Flag"]=df["Churn"].map({"Yes":1,"No":0})
```

```python
df.info()
```

**As we can see above, the Last_Interaction_Date column still has plaenty missing values. To correct it, we will add mean values as we did earlier.**

```python
df["Last_Interaction_Date"] = df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median())
```

**Let's recheck it.**

```python
df.info()
```

**As the final step, we will export the cleaned data for further analysis.
Exporting the cleaned DataFrame to a new CSV file**

```python
df.to_csv("Clean_Churn_Data.csv", index=False)
```
